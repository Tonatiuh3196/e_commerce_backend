```python
BASE_URL = "http://localhost:8001"

shoppingCart = {
    "user_id": 1,
    "items": [
        {"product_id": 1, "quantity": 2},
        {"product_id": 2, "quantity": 3}
    ]
}

# 🔹 Crear el carrito primero (si es necesario)
response_cart = requests.post(f"{BASE_URL}/shoppingCart/", json={"user_id": shoppingCart["user_id"]})

# Verificar si la creación fue exitosa
if response_cart.status_code == 201:
    cart_data = response_cart.json()
    print("Respuesta del carrito:", cart_data)  # Imprime la respuesta para asegurarte de que contiene el cart_id
    if "id" in cart_data:
        cart_id = cart_data["id"]  # Asegurarse de que 'id' está presente
        print(f"Carrito creado con ID: {cart_id}")
    else:
        print("No se encontró 'id' en la respuesta del carrito.")
        exit()  # Detener si no se recibe el 'id'
else:
    print("Error al crear el carrito:", response_cart.text)
    exit()  # Detener la ejecución si el carrito no se crea correctamente

# 🔹 Agregar productos al carrito
for item in shoppingCart["items"]:
    item["cart_id"] = cart_id  # Agregamos el ID del carrito a cada ítem
    response = requests.post(f"{BASE_URL}/shoppingCart/items/", json=item)
    
    print(response.status_code)
    print(response.text)
```
